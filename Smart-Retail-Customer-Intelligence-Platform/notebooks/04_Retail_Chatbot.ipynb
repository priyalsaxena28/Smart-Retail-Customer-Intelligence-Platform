```python
from sentence_transformers import SentenceTransformer
import numpy as np
import pickle
```

```python
questions = [
    "What is your return policy?",
    "How many days can I return a product?",
    "Do you provide home delivery?",
    "What are your store timings?",
    "Do you have cash on delivery?",
    "How can I track my order?",
    "Do you offer discounts?",
    "Can I exchange my product?",
    "How do I contact customer support?",
    "What payment methods do you accept?"
]

answers = [
    "Products can be returned within 30 days with the original receipt.",
    "Products can be returned within 30 days.",
    "Yes, home delivery is available for eligible locations.",
    "Our store is open from 9 AM to 9 PM every day.",
    "Yes, Cash on Delivery is available for selected locations.",
    "You can track your order from the Orders section of your account.",
    "Yes, discounts are available during seasonal sales and promotions.",
    "Yes, products can be exchanged within 30 days if eligible.",
    "You can contact customer support through email or phone.",
    "We accept UPI, credit cards, debit cards, net banking, and cash."
]
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
question_embeddings = model.encode(questions)

print("Knowledge base created successfully!")
```

```python
def chatbot(query):

    query_embedding = model.encode([query])

    scores = np.dot(question_embeddings, query_embedding.T).flatten()

    best_match = np.argmax(scores)

    return answers[best_match]
```

```python
print(chatbot("Can I return my order?"))

print(chatbot("Is cash on delivery available?"))

print(chatbot("When does your shop open?"))
```

```python
with open("chatbot.pkl", "wb") as f:
    pickle.dump(
        {
            "questions": questions,
            "answers": answers,
            "embeddings": question_embeddings
        },
        f
    )

print("Chatbot saved successfully!")
```
